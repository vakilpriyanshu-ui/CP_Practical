In a liquid phase reactor the reaction A+B --> C reaches equilibrium. Initial concentrations are [A] = 1.0 M, [B] = 1.0 M, and [C] = 0.0 mol/L. If x mol/L C is formed Kc = x/(1-x)(1x) = x/(1-x)^2., which rearranges to the quadratic equation Kc*xx-(2Kc+1)x+Kc=0. Write a python code that:
 
 a)reads Kc using input(), prints the data type of he raw input and the converted value, and shows a clear messege if the text is not a number or if Kc<=0 

 b)find a,b,c the disciminant and both roots 
 
 c)selects the physically meaningfull root (0<=x<=1) 
 
 d)prints equilibrium A,B,C and the conversion of the limiting reactant = x/[A]*100(%) 
 
 e)classifies the conversion using if/elif/else >= 70% "high",>= 40% "moderate", otherwise "low" Test with Kc=4, Kc=0.5 Kc=abc Kc=-2

```python
import math

raw_input = input("Enter a positive equilibrium constant Kc: ")
print("Raw input type:", type(raw_input))

try:
    Kc = float(raw_input)
except ValueError:
    print("Error: please enter a numeric value for Kc.")
else:
    print("Converted Kc:", Kc)

    if not math.isfinite(Kc) or Kc <= 0:
        print("Error: Kc must be a positive, finite number.")
    else:
        # Kc*x^2 - (2*Kc + 1)*x + Kc = 0
        a = Kc
        b = -(2 * Kc + 1)
        c = Kc
        discriminant = b**2 - 4 * a * c

        root_1 = (-b + math.sqrt(discriminant)) / (2 * a)
        root_2 = (-b - math.sqrt(discriminant)) / (2 * a)

        print("a =", a)
        print("b =", b)
        print("c =", c)
        print("Discriminant =", discriminant)
        print("Root 1 =", root_1)
        print("Root 2 =", root_2)

        physical_roots = [root for root in (root_1, root_2) if 0 <= root <= 1]
        if not physical_roots:
            print("Error: no physically meaningful root was found.")
        else:
            x = physical_roots[0]
            equilibrium_A = 1.0 - x
            equilibrium_B = 1.0 - x
            equilibrium_C = x
            conversion = x / 1.0 * 100

            print("Physical root x =", x, "mol/L")
            print("Equilibrium [A] =", equilibrium_A, "mol/L")
            print("Equilibrium [B] =", equilibrium_B, "mol/L")
            print("Equilibrium [C] =", equilibrium_C, "mol/L")
            print("Limiting-reactant conversion =", conversion, "%")

            if conversion >= 70:
                category = "high"
            elif conversion >= 40:
                category = "moderate"
            else:
                category = "low"

            print("Conversion category:", category)
```

```python
from math import exp

# Part A: Batch reactor simulation
C0 = 2.0
k = 0.08
dt = 1.0
total_time = 10

concentration = C0

print("Time (min)\tConcentration (mol/L)\tConversion (%)")
for time in range(total_time + 1):
    conversion = (C0 - concentration) / C0 * 100
    print(f"{time}\t\t{concentration:.4f}\t\t\t{conversion:.2f}")

    if time < total_time:
        concentration = concentration - k * concentration * dt

C_euler = concentration
C_exact = C0 * exp(-k * total_time)
absolute_error = abs(C_euler - C_exact)

print(f"\nEuler concentration at 10 min: {C_euler:.6f} mol/L")
print(f"Exact concentration at 10 min: {C_exact:.6f} mol/L")
print(f"Absolute error: {absolute_error:.6f} mol/L")

# Continue until conversion reaches 60% or time reaches 100 min
time = 0
concentration = C0

while (C0 - concentration) / C0 < 0.60 and time < 100:
    concentration = concentration - k * concentration * dt
    time += dt

print(f"\nTime needed: {time:g} min")
print(f"Final concentration: {concentration:.6f} mol/L")

# Part B: Batch label digit count and sum
label = input("\nEnter a batch label: ")
print('----------------------------------------------------------------')
print(label)
print('----------------------------------------------------------------')


print("Label type:", type(label))

if label == "":
    print("The label is empty.")
else:
    digit_count = 0
    digit_sum = 0

    for character in label:
        if "0" <= character <= "9":
            digit_count += 1
            digit_sum += int(character)

    if digit_count == 0:
        print("The label contains no digits.")
    else:
        print(f"Number of digits: {digit_count}")
        print(f"Sum of digits: {digit_sum}")
```

PartA. A laboratory has n candidate additives and wants to test r of them togrther in one formulation (order does not matter). Write a recursive function factorial(n) (remember the base case for 0 ) and a function combination (n,r)=n!/(r!(n-r)) that uses it. Reject invalid input (negative values, r>n ). Print the number of possible formulations for n=6, r=3 and show the messege for n=5, r=7

 PartB. During a reaction, samples are taken at times = Fibonacci number*2. Write an Iterative function fib_iter(n) that returns a list of a first n fibonacci numbers and a recursive function fib_rec(k) that returns the k-th fibonacci number (F=0, F1=1) for n=10, check that both functions agree and print a table: Sample no. | Fibonacci tag | Sampling time (min)

```python
# Part A: Additive combinations
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative values.")
    if n == 0:  # Base case: 0! = 1
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if n < 0 or r < 0 or r > n:
        raise ValueError("Invalid input: require n >= 0 and 0 <= r <= n.")
    return factorial(n) // (factorial(r) * factorial(n - r))


print("Possible formulations for n=6, r=3:", combination(6, 3))

try:
    print("Possible formulations for n=5, r=7:", combination(5, 7))
except ValueError as error:
    print(error)


# Part B: Fibonacci sampling schedule
def fib_iter(n):
    """Return a list containing the first n Fibonacci numbers."""
    if n < 0:
        raise ValueError("n must not be negative.")

    numbers = []
    a, b = 0, 1
    for _ in range(n):
        numbers.append(a)
        a, b = b, a + b
    return numbers


def fib_rec(k):
    """Return Fibonacci number F_k, where F_0=0 and F_1=1."""
    if k < 0:
        raise ValueError("k must not be negative.")
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib_rec(k - 1) + fib_rec(k - 2)


n = 10
fibonacci_tags = fib_iter(n)
recursive_tags = [fib_rec(k) for k in range(n)]

print("\nIterative and recursive results agree:", fibonacci_tags == recursive_tags)
print("\n   Sample no.   Fibonacci tag   Sampling time (min)")
for sample_no, tag in enumerate(fibonacci_tags, start=1):
    print(f"{sample_no:10}  {tag:13}  {tag * 2:19}")
```

Q.4. Kinetics and gas data: Pandas and matplotlib (10 marks) hydrolysis of an ester was studied at two temperatures. the concentration (mol/L) was recorded: Time = (0,10,20,30,40,50) Run1_mol_L (Low T) = (1,00, 0.74, 0.55, 0.41, 0.30, 0.22) Run2-mol_L (high T) = (1.00, 0.61, 0.37, 0.22, 0.14, 0.08) a) Create a dataFrame from this table. Save it as kinetics.csv (without the index column) and as kinetics.json. Read the CSV back and print its shape and datatypes. b)Print the mean and standard deviation of Run1 and Run2 and state which run shows more vabration. c)Plot both runs on the graph (Time on X-axis) with title, axis labels with units, legend and grid. d)Nitrogen gas is compressed slowly in a piston cylinder at constant temperature: volume_L = (1,2,3,4,5) pressure_bar = (4.00, 2.03, 1.31, 1.02, 0.81) Plot pressure vs volume as a scater plot. Calculate k as the mean of P*V and drAw the reference curve p = k/V on the same graph. Add title, labels with units and legend.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Part A: Ester hydrolysis data
kinetics = pd.DataFrame({
    "Time_min": [0, 10, 20, 30, 40, 50],
    "Run1_mol_L": [1.00, 0.74, 0.55, 0.41, 0.30, 0.22],
    "Run2_mol_L": [1.00, 0.61, 0.37, 0.22, 0.14, 0.08],
})

kinetics.to_csv("kinetics.csv", index=False)
kinetics.to_json("kinetics.json", orient="records", indent=4)

# Read the CSV back and inspect it
kinetics_read = pd.read_csv("kinetics.csv")
print("CSV shape:", kinetics_read.shape)
print("Data types:\n", kinetics_read.dtypes)

# Print summary statistics
for run in ["Run1_mol_L", "Run2_mol_L"]:
    print(f"\n{run}:")
    print(f"Mean: {kinetics_read[run].mean():.4f} mol/L")
    print(f"Standard deviation: {kinetics_read[run].std():.4f} mol/L")

if kinetics_read["Run1_mol_L"].std() > kinetics_read["Run2_mol_L"].std():
    print("\nRun 1 shows more variation (higher standard deviation).")
else:
    print("\nRun 2 shows more variation (higher standard deviation).")

# Plot both concentration runs
plt.figure()
plt.plot(kinetics_read["Time_min"], kinetics_read["Run1_mol_L"],
         marker="o", label="Run 1 (Low T)")
plt.plot(kinetics_read["Time_min"], kinetics_read["Run2_mol_L"],
         marker="s", label="Run 2 (High T)")
plt.title("Ester Hydrolysis at Two Temperatures")
plt.xlabel("Time (min)")
plt.ylabel("Concentration (mol/L)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()


# Part D: Nitrogen pressure-volume data
volume_L = np.array([1, 2, 3, 4, 5], dtype=float)
pressure_bar = np.array([4.00, 2.03, 1.31, 1.02, 0.81])

k = np.mean(pressure_bar * volume_L)
print(f"\nMean P*V constant, k: {k:.3f} bar·L")

# Smooth reference curve: P = k / V
volume_curve = np.linspace(volume_L.min(), volume_L.max(), 200)
pressure_curve = k / volume_curve

plt.figure()
plt.scatter(volume_L, pressure_bar, label="Measured data")
plt.plot(volume_curve, pressure_curve, label=f"Reference curve: P = {k:.3f}/V")
plt.title("Nitrogen Gas: Pressure vs Volume")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
